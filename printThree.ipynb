```python
#basic data types:
a=10
print(type(a))
b=3.14
print(type(b))
c="Python"
print(type(c))
d=True
print(type(d))
e=False 
print(type(e))
f=2+3j # complex data types consist of real and imaginary parts
print(type(f))
#extract real and imaginary part from comlpex data 
print(f"Real part: {f.real}")
print(f"imaginary part: {f.imag}")
```

```python
#special data types:
#list: a sequence of elements declared with square bracket 
my_list=[1,2,3,4,5,6]
print(type(my_list))
```

```python
#tuple: declared in round bracket()
my_tuple=(1,2,3,4,5,6)
print(tuple(my_tuple))
```

```python
#set -> declared within curly brackets :
my_set={1,2,3,4,5,6}
print(type(my_set))
```

```python
#dictionary -> declared on a {key-value} pairs 
my_dict={"name":"Andrew","Age":"20","City":"New york"}
print(type(my_dict))
```

```python
# type conversion / type casting:
'''
Changing data type from one data type from another data type 
''' 
a=10
print(type(a))
a_float=float(a)
print(type(a_float))
a_str=str(a)
print(type(a_str))
a_bool=bool(a)
print(type(a_bool))
```

```python
# operators in python 
# arithemetric operators 
a=10
b=20
print(a+b) # addition 
print(a-b) # Subtraction
print(a*b) # multiplication
print(a/b) # division 
print(a//b) # floor division (quatient)
print(a%b) # Modulo (remainder)
print(a**b) #exponential (to power of)
```

```python
# assignment operator 
x=10
x=x+3
print(x)


x+=3
print(x)
x-=3
print(x)
x*=3
print(x)
x/=3
print(x)
x//=3
print(x)
x%=3
print(x)
x**=3
print(x)
```

```python
# comparison operator 
x=20
y=30
print(x>y) # greater than 
print(x>=y) # greater than and equal to
print(x<y) # lesser than 
print(x<=y) # lesser than and equal to 
print(x==y) # x is equal to y 
print(x!=y) # x is not equal to y 
```

```python
# logical operator 
a=100
b=150 
c=50
print(a>b and a>c) # returns true when both the conditions are true else return false 
print(a>b or a>c)  # returns true when atleast one condition is true 
print(not(a>b))    # returns opposite of actual output 
```

```python
# identity operator : is, is not 
a=10 
print(id(a))
b=10
print(id(b))
print(a is b)
print(a is not b)
```

```python
# membership operator : in, not in 
a=[1,2,3,4,5]
print(4 in a) # returns true is 4 is present in the list 'a', else false 
print(10 not in a) # returns true is 10 is not present in the list 'a', else false 
```
