```python
# Item 1
item1_name = input("Enter the Item name: ")
item1_price = int(input("Enter the price {item1_name}:₹ "))
item1_qty = int(input("Enter the qty{item1_name}: "))

 #item total
item1_total=item1_price*item1_qty

# Item 2
item2_name = input("Enter the Item name: ")
item2_price = int(input("Enter the price {item2_name}:₹ "))
item2_qty = int(input("Enter the qty{item2_name}: "))

 #item total
item2_total=item2_price*item2_qty

# Item 3
item3_name = input("Enter the Item name: ")
item3_price = int(input(f"Enter the price {item3_name}:₹ "))
item3_qty = int(input("Enter the qty{item3_name}: "))

 #item total
item3_total=item3_price*item3_qty

#subtotal:
subtotal = item1_total+item2_total+item3_total

# calculate 10% discount:
discount_amount = subtotal*0.10

#Total_bill_amount 
total_bill_amount= subtotal - discount_amount

#calculate GST (12%):
gst_tax_amount = total_bill_amount*0.12

#get the grand total :
grand_total = total_bill_amount+gst_tax_amount



print("=="*30)
print("            RECEIPT             ")
print("=="*30)
print(f"(item1_name) x (item1_qty): (item1_total:.2f)")
print(f"(item2_name) x (item2_qty): (item2_total:.2f)")
print(f"(item3_name) x (item3_qty): (item3_total:.2f)")
print("-"*30)
print(f"Subtotal Bill Amount: (subtotal:.2f)")
print(f"10% discount:-(discount_amount:.2f)")
print("_"*30)
print(f"Total Bill Amount: ${total_bill_amount: .2f}")
print(f"GST(12%): +${gst_tax_amount:.2f}")
print("_"*30)
print(f"Grand total : ${grand_total:.2f}")
```

```python
# Conditional statement 
# if statement 
# if - else statement
# if - elif - else statement 
# nested - if statement
```

```python
''' 
if statement syntax : 
if condition: 
    # Value if the condition is true 
'''

age=int(input("Enter the age"))
if age>=18:
    print("The person is an Adult")
```

```python
''' 
if - else syntax:-
if condition:
    # value if the condition is true 
else:
    # value if the condition is false 
'''

age=int(input("Enter the age:"))
if age>=18:
    print("The person is adult")
else:
    print("The person is too young")
```

```python
''' 
if - elif - else condition 
used to check multiple condition
'''

mark=int(input("Enter the student mark: "))
if mark>=95:
    print("Grade A")
elif mark>=75:
    print("Grade B")
elif mark>=50:
    print("Grade C")
else:
    print("Grade F")
```

```python
'''
Nested - if:
Multiple if-else Statements

Syntax:
if condition:
    if condition:
        # value if condition is true
    else:
        # value if condition is false
else:
    # value if condition is false 
'''

print("Train Ticket booking System")
age=int(input("Enter the age of the passenger"))
membership_card=input("Say (yes/no)")

if age>=18:
    if membership_card=="yes":
        print("The ticket fare is ₹150")
    else:
        print("The ticket fare is ₹200")
else:
    if membership_card=="yes":
        print("The ticket fare is ₹80")
    else:
        print("The ticket fare is ₹120")
        
```
