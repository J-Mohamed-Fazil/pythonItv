```python
name = "Anand"
age=20
city="Chennai"
print(name,age,city,sep=',')
```

```python
print("The name of the person",name,"he is",age,"years old from city",city)
```

```python
# f-string method
print(f"The name of the person is {name} and he is {age} years old from {city} city")
```

```python
# format method
print("The name of the person is {} and he is {} years old from {} city".format(name,age,city))
```

```python
# format method : position Value
print("The name of the person is {1} and he is {0} years old from {2} city".format(age,name,city))
```

```python
# format method : keyword Value
print("The name of the person is {name} and he is {age} years old from {city} city".format(age=30,name="Andrew",city="Mumbai"))
```
